```python
import socket

s = socket.socket()

host = "127.0.0.1"
port = 8080

s.connect((host, port))

print("Connected to chat server")

while True:

    incoming_message = s.recv(1024).decode()

    print("Server :", incoming_message)
    print()

    message = input(">> ")

    s.send(message.encode())

    print("Sent")
    print()
```
